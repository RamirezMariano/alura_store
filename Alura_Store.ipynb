```python
import pandas as pd
import matplotlib.pyplot as plt
# estilo de gráficos
plt.style.use("seaborn-v0_8")

plt.rcParams["figure.figsize"] = (8,5)
plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.labelsize"] = 12
plt.rcParams["figure.dpi"] = 120
#Cargar los archivos de tiendas
url1 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda1 = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda1.head()
```

```python
#Verificar y comprobar los datos cargados
print("Tienda 1:", tienda1.shape)
print("Tienda 2:", tienda2.shape)
print("Tienda 3:", tienda3.shape)
print("Tienda 4:", tienda4.shape)
```

**1- Análisis de Ventas**

En este paso se analizaron las ventas por categoría de producto en cada tienda.
Para ello, los datos fueron agrupados utilizando la columna Categoría del Producto, lo que permitió contar cuántas ventas se realizaron en cada categoría.

Dicho análisis permite identificar cuáles son las categorías más populares en cada tienda y comprender mejor el comportamiento de compra de los clientes.


```python
ventas_categoria_t1 = tienda1.groupby("Categoría del Producto").size()
ventas_categoria_t2 = tienda2.groupby("Categoría del Producto").size()
ventas_categoria_t3 = tienda3.groupby("Categoría del Producto").size()
ventas_categoria_t4 = tienda4.groupby("Categoría del Producto").size()

print("Tienda 1")
print(ventas_categoria_t1)

print("\nTienda 2")
print(ventas_categoria_t2)

print("\nTienda 3")
print(ventas_categoria_t3)

print("\nTienda 4")
print(ventas_categoria_t4)
```

Visualización de ventas por categoría

Para comprender mejor la distribución de las ventas, se generó un gráfico de barras que muestra la cantidad de productos vendidos en cada categoría. Esta visualización permite identificar rápidamente qué tipos de productos tienen mayor demanda en cada tienda.

```python
fig, axs = plt.subplots(2,2, figsize=(12,10))

ventas_categoria_t1.plot(kind='bar', ax=axs[0,0], title="Tienda 1")
ventas_categoria_t2.plot(kind='bar', ax=axs[0,1], title="Tienda 2")
ventas_categoria_t3.plot(kind='bar', ax=axs[1,0], title="Tienda 3")
ventas_categoria_t4.plot(kind='bar', ax=axs[1,1], title="Tienda 4")

plt.tight_layout()
plt.show()
```

**2- Análisis de la calificación promedio**

En este paso se calculó la calificación promedio de los clientes para cada tienda utilizando la columna Calificación. Este indicador permite evaluar el nivel de satisfacción de los clientes con los productos y servicios ofrecidos por cada tienda.

A través de este análisis es posible identificar qué tiendas presentan mejores evaluaciones por parte de los consumidores.

```python
valoracion_t1 = tienda1["Calificación"].mean()
valoracion_t2 = tienda2["Calificación"].mean()
valoracion_t3 = tienda3["Calificación"].mean()
valoracion_t4 = tienda4["Calificación"].mean()

valoraciones = {
    "Tienda 1": valoracion_t1,
    "Tienda 2": valoracion_t2,
    "Tienda 3": valoracion_t3,
    "Tienda 4": valoracion_t4
}

print(valoraciones)
```

```python
plt.figure(figsize=(8,5))

plt.bar(valoraciones.keys(), valoraciones.values())

plt.title("Calificación promedio por tienda")
plt.xlabel("Tiendas")
plt.ylabel("Calificación promedio")

plt.ylim(3.9,4.1)

plt.show()
```

3- Análisis de productos más y menos vendidos

En este paso se analizaron los productos más y menos vendidos en cada tienda. Para eso, se utiliza la función value_counts sobre la columna Producto, eso permite identificar cuántas veces fue vendido cada artículo
Este ánalisis permite detectar la demanda y rotación de los productos

```python
for i, tienda in enumerate([tienda1, tienda2, tienda3, tienda4], start=1):

    ventas = tienda["Producto"].value_counts()

    mas_vendido = ventas.idxmax()
    menos_vendido = ventas.idxmin()

    print(f"Tienda {i}")
    print("Más vendido:", mas_vendido)
    print("Menos vendido:", menos_vendido)
    print()
```

```python
top10 = tienda1["Producto"].value_counts().head(10)

plt.figure(figsize=(8,5))
top10.plot(kind="bar")

plt.title("Top 10 productos más vendidos - Tienda 1")
plt.xlabel("Producto")
plt.ylabel("Cantidad de ventas")

plt.xticks(rotation=45)

plt.show()
```

```python
top10 = tienda2["Producto"].value_counts().head(10)

plt.figure(figsize=(8,5))
top10.plot(kind="bar")

plt.title("Top 10 productos más vendidos - Tienda 2")
plt.xlabel("Producto")
plt.ylabel("Cantidad de ventas")

plt.xticks(rotation=45)

plt.show()
```

```python
top10 = tienda3["Producto"].value_counts().head(10)

plt.figure(figsize=(8,5))
top10.plot(kind="bar")

plt.title("Top 10 productos más vendidos - Tienda 3")
plt.xlabel("Producto")
plt.ylabel("Cantidad de ventas")

plt.xticks(rotation=45)

plt.show()
```

```python
top10 = tienda4["Producto"].value_counts().head(10)

plt.figure(figsize=(8,5))
top10.plot(kind="bar")

plt.title("Top 10 productos más vendidos - Tienda 4")
plt.xlabel("Producto")
plt.ylabel("Cantidad de ventas")

plt.xticks(rotation=45)

plt.show()
```

**Productos más vendidos **

Para comprender mejor el comportamiento general de las ventas, se unificaron los datos de las cuatro tiendas en un único conjunto de datos. A partir de este dataset combinado, se identificaron los productos más vendidos en toda la cadena.

Esta visualización permite identificar qué productos tienen mayor demanda en el conjunto de las tiendas y cuáles representan una mayor proporción de las ventas.

```python
todas_tiendas = pd.concat([tienda1, tienda2, tienda3, tienda4])

top_productos = todas_tiendas["Producto"].value_counts().head(10)

plt.figure(figsize=(8,5))
top_productos.plot(kind="bar")

plt.title("Top 10 productos más vendidos en todas las tiendas")
plt.xlabel("Producto")
plt.ylabel("Cantidad de ventas")

plt.xticks(rotation=45)

plt.show()
```

**4- Envío Promedio por tienda**

En este paso se calculó el costo promedio de envío para cada tienda utilizando la columna Costo de envío. Este indicador permite comprender cuánto pagan los clientes en promedio por el envío de sus compras.

El análisis del costo de envío es relevante porque puede influir en la decisión de compra de los clientes y afectar el volumen de ventas de cada tienda.

```python
envio_t1 = tienda1["Costo de envío"].mean()
envio_t2 = tienda2["Costo de envío"].mean()
envio_t3 = tienda3["Costo de envío"].mean()
envio_t4 = tienda4["Costo de envío"].mean()

envios = {
    "Tienda 1": envio_t1,
    "Tienda 2": envio_t2,
    "Tienda 3": envio_t3,
    "Tienda 4": envio_t4
}

print(envios)
```

```python
plt.figure(figsize=(8,5))

barras = plt.bar(envios.keys(), envios.values())

plt.title("Costo de envío promedio por tienda")
plt.xlabel("Tiendas")
plt.ylabel("Costo de envío promedio")

for barra in barras:
    altura = int(barra.get_height())

    numero_vertical = "\n".join(str(altura))  # separa cada dígito

    plt.text(
        barra.get_x() + barra.get_width()/2,
        altura/2,
        numero_vertical,
        ha='center',
        va='center',
        color='white',
        fontsize=20
    )

plt.show()
```
